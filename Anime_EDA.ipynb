```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("anime_dataset_2000-2024.csv")
df
```

```python
studio_hits = (
    df.groupby("studio")
      .agg(
          hit_rate=("is_breakout_hit", "mean"),  # <-- FIXED
          show_count=("title", "count")
      )
      .query("show_count >= 3")
      .sort_values("hit_rate", ascending=False)
)

top_10 = studio_hits.head(10)

sns.barplot(
    x=top_10["hit_rate"] * 100,
    y=top_10.index
)

plt.xlabel("Breakout Hit Rate (%)")
plt.ylabel("Studio")
plt.title("Top 10 Studios by Breakout Hit Rate")
plt.show()
```

```python

sns.set_style("whitegrid")



# ============================================================
# 1. Score distribution
# ============================================================

fig, axes = plt.subplots(3, 2, figsize=(14, 16))

sns.histplot(
    data=df,
    x="final_score",
    bins=30,
    ax=axes[0, 0]
)

axes[0, 0].set_title("Final Score Distribution")
axes[0, 0].set_xlabel("Final Score")


# ============================================================
# 2. Breakout rate by source
# ============================================================

top_src = df["source"].value_counts().head(8).index

source_subset = df[df["source"].isin(top_src)]

sns.barplot(
    data=source_subset,
    x="source",
    y="is_breakout_hit",
    ax=axes[0, 1],
    estimator="mean"
)

axes[0, 1].set_title("Breakout Rate by Source")
axes[0, 1].set_ylabel("Breakout Rate")
axes[0, 1].tick_params(axis="x", rotation=45)


# ============================================================
# 3. Studio history vs final score
# ============================================================

sns.scatterplot(
    data=df,
    x="studio_hist_avg_score",
    y="final_score",
    alpha=0.4,
    ax=axes[1, 0]
)

axes[1, 0].set_title("Studio Historical Score vs Final Score")
axes[1, 0].set_xlabel("Studio Historical Average Score")
axes[1, 0].set_ylabel("Final Score")


# ============================================================
# 4. Debut vs experienced director
# ============================================================

sns.barplot(
    data=df,
    x="is_debut_director",
    y="is_breakout_hit",
    ax=axes[1, 1]
)

axes[1, 1].set_title("Breakout Rate: Debut vs Experienced Director")
axes[1, 1].set_ylabel("Breakout Rate")
axes[1, 1].set_xticks([0, 1])
axes[1, 1].set_xticklabels(["Experienced", "Debut"])


# ============================================================
# 5. Final score by season
# ============================================================

sns.boxplot(
    data=df,
    x="season",
    y="final_score",
    ax=axes[2, 0]
)

axes[2, 0].set_title("Final Score by Season")
axes[2, 0].set_xlabel("Season")
axes[2, 0].set_ylabel("Final Score")


# ============================================================
# 6. Breakout rate by year
# ============================================================

yearly = (
    df.groupby("season_year")["is_breakout_hit"]
    .mean()
)

axes[2, 1].plot(
    yearly.index,
    yearly.values,
    marker="o"
)

axes[2, 1].set_title("Breakout Rate by Year")
axes[2, 1].set_xlabel("Year")
axes[2, 1].set_ylabel("Breakout Rate")


# ============================================================
# Layout + save
# ============================================================

plt.tight_layout()

plt.savefig(
    "eda_overview.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# Correlations with final score
# ============================================================

numeric_cols = [
    "final_score",
    "studio_hist_avg_score",
    "director_hist_avg_score",
    "studio_hist_show_count",
    "director_hist_show_count",
    "episodes",
    "duration",
    "season_year"
]

correlations = (
    df[numeric_cols]
    .corr()["final_score"]
    .sort_values(ascending=False)
)

print("\nCorrelation with final_score:")
print(correlations)
```

```python

sns.set_style('whitegrid')


# popularity = total_favorites (audience reach), not score
dir_stats = df.groupby('director').agg(
    total_favorites=('final_favorites', 'sum'),
    show_count=('title', 'count')
).query('show_count >= 2')

studio_stats = df.groupby('studio').agg(
    total_favorites=('final_favorites', 'sum'),
    show_count=('title', 'count')
).query('show_count >= 2')

top_dir = dir_stats.sort_values('total_favorites', ascending=False).head(10)
top_studio = studio_stats.sort_values('total_favorites', ascending=False).head(10)

top_shows = df[['title', 'final_favorites']].sort_values('final_favorites', ascending=False).head(10)

fig, axes = plt.subplots(1, 3, figsize=(20, 6))

sns.barplot(x=top_dir['total_favorites'], y=top_dir.index, ax=axes[0], color='seagreen')
axes[0].set_title('Most Popular Directors (total favorites)')

sns.barplot(x=top_studio['total_favorites'], y=top_studio.index, ax=axes[1], color='steelblue')
axes[1].set_title('Most Popular Studios (total favorites)')

sns.barplot(x=top_shows['final_favorites'], y=top_shows['title'], ax=axes[2], color='coral')
axes[2].set_title('Most Popular Shows (favorites)')

plt.tight_layout()
plt.savefig('popularity_rankings.png', dpi=150)
```
